```python
!pip install gensim
import gensim.downloader as api
import random
```

```python
word_vectors = api.load("word2vec-google-news-300")
```

```python
def create_story(seed_word):
    similar_words = [word for word, _ in word_vectors.most_similar(seed_word, topn=5)]
    story = f"Once upon a time, there was a {seed_word}. It lived in a world filled with "
    story += ', '.join(similar_words) + "."
    story += f" Every day, the {seed_word} would explore new places, meeting {random.choice(similar_words)} and "
    story += f"learning about {random.choice(similar_words)}. It was an adventure every day!"
    return story
```

```python
seed_word = 'adventure'
story = create_story(seed_word)
```
