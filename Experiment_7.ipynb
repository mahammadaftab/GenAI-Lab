```python
from transformers import pipeline
```

```python
summarizer = pipeline("text-generation", model="sshleifer/distilbart-cnn-12-6")
```

```python
long_text = """
The process of machine learning involves training algorithms on data to allow them to make predictions or decisions based on new data.
It begins with gathering large sets of data and then using this data to train models. The models learn patterns and relationships
within the data, which can be used to make informed predictions when new data is presented. Machine learning is used in various
fields including healthcare, finance, and technology, where it aids in tasks such as medical diagnosis, fraud detection, and
customer service automation. Over time, machine learning models continue to improve as they are exposed to more data, becoming
increasingly accurate and effective at their tasks.
"""
```

```python
summary = summarizer(long_text, max_length=100, min_length=30, do_sample=False)
```

```python
print("Original Text:\n", long_text)
print("\nSummarized Text:\n", summary[0]['generated_text'])
```
