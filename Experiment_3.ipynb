```python
!pip install gensim
import gensim
from gensim.models import Word2Vec
import nltk
from nltk.tokenize import word_tokenize
```

```python
medical_corpus=[
    "The doctor diagnosed the paient wirh a serve infection."
    "The hospital provides healthcare service to patients."
    "The nurse assosted the doctor during the surgey."
    "Doctors often recommed regular checkups to prevent illness."
    "The surgeon performed a complex operation to save the patients life."
    "Hospitals are equipped with advanced medical technology."
    "vaccines help to prevent infectious diseases."
    "Medical research is crucial for the development of new treatments."
    "A medical emergency requires immediate attention from healthcare professionals."
]
nltk.download('punkt_tab')
```

```python
tokenized_corpus=[word_tokenize(sentence.lower()) for sentence in medical_corpus]
```

```python
model=Word2Vec(sentences=tokenized_corpus,vector_size=100,window=5,min_count=1,workers=4)
```

```python
model.save("medical_Word2Vec.model")
print("Model saved as 'medical_Word2Vec.model.'")
```

```python
def find_similar_words(input_word,topn=5):
  """
  Given a word, find the most semantically similar words in the model.
  :param input_word: The word for which to find similar words.
  :param topn: number of similar words to return.
  :return:List of tuples containing similar words and their simicarity scores
  """
  if input_word in model.wv:
    similar_words=model.wv.most_similar(input_word,topn=topn)
    return similar_words
  else:
    return f"word'{input_word}'not found in vocaburlary."
```

```python
similar_words_to_doctor=find_similar_words("doctor",topn=5)
print("\n Top 5 semantically similar words to 'doctor':")
for word,similarity_score in similar_words_to_doctor:
  print(f"{word}:{similarity_score}")
```
