```python
from transformers import pipeline
```

```python
sentiment_analyzer = pipeline("sentiment-analysis")
```

```python
reviews = [
    "The product is absolutely fantastic! I love it.",
    "I'm not satisfied with the quality of this product.",
    "Great Customers service, would definitely recommand!",
    "This is the worst purchase I've ever made. Totally regret it.",
    "I feel neutral about the product, it's okay, not great but not bad either.",
]
```

```python
for review in reviews:
  result=sentiment_analyzer(review)
  sentiment=result[0]['label']
  confidence=result[0]['score']
  print(f"Review:{review}\nSentiment:{sentiment}(confidence:{confidence:4f})\n")
```
